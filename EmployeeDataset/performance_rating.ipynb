```python
import pandas as pd
df = pd.read_csv('employee_data-selected-columns.csv')
import numpy as np
```

```python
employee_data = pd.read_csv('employee_data.csv')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv ('employee_data-selected-columns.csv')

performance_data = df["Performance Score"]   
employee_rating = df["Current Employee Rating"]                      
colors = ["#1f77b4", "#ff7f0e", "#2ca02c", "#d62728", "#8c564b"]    

plt.pie(employee_rating , labels=performance_data , colors=colors,
autopct='%1.1f%%', shadow=True, startangle=140)

plt.show()
```

```python
performance_counts = pd.read_csv("employee_data-selected-columns.csv")[
    "Performance Score"
].value_counts()

fig, ax = plt.subplots()
wedges, _, _ = ax.pie(
    performance_counts,
    autopct="%1.1f%%",
    startangle=140,
    colors=colors[:len(performance_counts)]
)
ax.legend(wedges, performance_counts.index, title="Performance Score",
          loc="center left", bbox_to_anchor=(1, 0.5))
ax.set_title("Performance Score Distribution")
ax.axis("equal")
plt.tight_layout()
plt.show()
```

```python
df = pd.read_csv("employee_data-selected-columns.csv")

rating_counts = df["Current Employee Rating"].value_counts().sort_index()
plt.pie(
    rating_counts,
    labels=rating_counts.index,
    autopct="%1.1f%%",
    startangle=140
)
plt.title("Current Employee Rating")
plt.axis("equal")
plt.show()
```

```python
df = pd.read_csv("employee_data-selected-columns.csv")
rating_counts = df["Current Employee Rating"].value_counts().sort_index()

fig, ax = plt.subplots()
wedges, _, _ = ax.pie(
    rating_counts,
    autopct="%1.1f%%",
    startangle=140,
    colors=colors[:len(rating_counts)]
)
ax.legend(
    wedges,
    rating_counts.index,
    title="Current Employee Rating",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)
ax.set_title("Current Employee Rating Distribution")
ax.axis("equal")
plt.tight_layout()
plt.show()
```
